#### 🧩 Concept card: `from ... import ...`

| | |
|---|---|
| **Syntax** | `from library import part`<br>`from library import part as nickname`<br>`import library.part as nickname` |
| **What it is** | Takes just one part out of a library, so you can use it by its short name. |
| **Where it's used in ML** | Grabbing the exact pieces you need for building and training a model, without typing the long path every time. |
| **Real example** | `from tensorflow import keras` (the model-building part of TensorFlow), `from sklearn.model_selection import train_test_split` (splits data into practice and test sets). |

```python
from matplotlib import pyplot as plt
```

Most people write the same thing in this shorter way. Both lines do exactly the same job:

```python
import matplotlib.pyplot as plt
```

| Word | What it is | Cricket idea | Example |
|---|---|---|---|
| **Library** | A big collection of code someone else wrote | The whole kit bag | `tensorflow`, `matplotlib` |
| **Module** | One part inside a library | One pocket in the bag | `pyplot` |
| **Function** | A tool that does one job when you call it | The bat: you swing it, something happens | `plt.bar( )`, `tf.constant( )` |
| **Class** | A blueprint that makes objects | The jersey design: every player's shirt is made from it | `tf.Variable` |

So: **`matplotlib` is a library, `pyplot` is a module (not a class, not a function), and `bar` is a function inside it.**

The dot `.` means "go inside". `plt.bar` = go inside `plt` and get `bar`.

- **Function** = a tool that does one job. Brackets `( )` mean "do it now".
- **Inputs** (also called *arguments*) = what you give it, inside the brackets, split by commas.
- **Result** (also called *return value*) = what comes back.
- **Method** = a function that belongs to an object, used with a dot: `object.method( )`.
- **No brackets** (like `runs.shape`) = not a function, just a fact stored on the object.

| Function | How to write it | What it does | Example → result |
|---|---|---|---|
| `inspect.ismodule` | `inspect.ismodule(x)` | Answers: is `x` a module? | `inspect.ismodule(plt)` → `True` |
| `inspect.isfunction` | `inspect.isfunction(x)` | Answers: is `x` a function? | `inspect.isfunction(plt.bar)` → `True` |
| `inspect.isclass` | `inspect.isclass(x)` | Answers: is `x` a class? | `inspect.isclass(tf.Variable)` → `True` |

#### 🧩 Concept card: Library · Module · Function · Class

| | |
|---|---|
| **Syntax** | `library.module.function(inputs)`<br>`object = ClassName(inputs)`<br>`object.method(inputs)` |
| **What it is** | Library = big collection of code. Module = one part of it. Function = a tool you call with `( )`. Class = a blueprint you use to make objects. A *method* is a function that belongs to an object. |
| **Where it's used in ML** | Building any model reads exactly like this: go inside TensorFlow → Keras → layers, then use a class to make a layer. |
| **Real example** | `tf.keras.layers.Dense(64)`: `tf` library → `keras` module → `layers` module → `Dense` class makes one layer object. Then `model.fit(data)` is a method: a function belonging to the model. |

You don't have to guess. Python can tell you what something is:

```python
import inspect

print(inspect.ismodule(plt))
print(inspect.isfunction(plt.bar))
print(inspect.isclass(tf.Variable))
```

Three `True`s: `plt` is a module, `plt.bar` is a function, `tf.Variable` is a class.

`inspect` is a small toolbox that comes with Python. It's only used here to check, you won't need it again.

#### 🧩 Concept card: `tf.stack`

| | |
|---|---|
| **Syntax** | `tf.stack([t1, t2, ...])`<br>`tf.stack([t1, t2, ...], axis=1)` |
| **What it is** | Puts tensors of the same shape on top of each other, making one new direction. `axis` says where the new direction goes. |
| **Where it's used in ML** | Making a **batch**. Models learn from many examples at once, so single examples are stacked together. Also used to join separate columns into one table. |
| **Real example** | 32 photos, each `(224, 224, 3)` → `tf.stack` → one batch `(32, 224, 224, 3)` given to the model in one go. Or height, weight and age lists → `tf.stack(..., axis=1)` → one table of people. |

#### 🧩 Concept card: `tf.reduce_sum` / `reduce_mean` / `reduce_max` / `reduce_min`

| | |
|---|---|
| **Syntax** | `tf.reduce_sum(t)`<br>`tf.reduce_mean(t)`<br>`tf.reduce_max(t, axis=0)` |
| **What it is** | Squeezes many numbers down to fewer: total, average, biggest, smallest. |
| **Where it's used in ML** | **Every model's training score is a `reduce_mean`.** The model's mistakes on all examples are averaged into one number called the *loss*, and training tries to make that number smaller. |
| **Real example** | Loss = `tf.reduce_mean((prediction - actual) ** 2)`. Accuracy = `tf.reduce_mean(correct)`. Both are single numbers you watch while training. |

|---|---|---|---|

| `v.assign_sub(x)` | Takes `x` away from what's there | old − x | Penalty runs taken off |

You can't use `=` to change it: `score = 10` would throw away the scoreboard and make `score` a plain number. Always use the methods.

```python
score.assign_sub(5)
print(score.numpy())
```

```python
score.assign(0)
print(score.numpy())
```

```python
fastest = int(tf.argmax(strike_rate))
colors = ["#94a3b8"] * 5
colors[fastest] = "#ea580c"

plt.bar(players, strike_rate.numpy(), color=colors)
plt.title("Fastest scorer: " + players[fastest])
plt.ylabel("Runs per 100 balls")
plt.show()
```

`["#94a3b8"] * 5` makes a list of 5 greys. Then box `fastest` is swapped to orange.

#### 🔧 Functions in this chapter

| Function | How to write it | What it does | Example → result |
|---|---|---|---|
| `bottom=` | `plt.bar(..., bottom=other_heights)` | An input to `plt.bar`: start each bar on top of another one. | → stacked bars |
| `label=` | `plt.bar(..., label="text")` | An input to `plt.bar`: the name shown in the legend. | → legend entry |

```python
names = ["Rohit", "Gill", "Kohli", "Iyer", "Rahul", "Hardik"]
card = tf.constant([
    [74, 58, 8, 3],
    [52, 49, 6, 1],
    [103, 87, 10, 2],
    [44, 40, 4, 1],
    [31, 27, 2, 1],
    [38, 20, 3, 3]
], dtype=tf.float32)
print(card.shape)
```

`dtype=tf.float32` makes everything decimals from the start, so you never need `tf.cast` later.

Each column is one stat. Slice them out (chapter 6):

```python
runs = card[:, 0]
balls = card[:, 1]
fours = card[:, 2]
sixes = card[:, 3]
```

Whole-tensor math (chapter 7). A four is worth 4 runs and a six is worth 6, the rest was running between the wickets:

```python
strike_rate = runs / balls * 100
boundary_runs = fours * 4 + sixes * 6
running_runs = runs - boundary_runs
print(strike_rate)
```

Summing up (chapter 8):

```python
print("Team total:", tf.reduce_sum(runs).numpy())
print("Top scorer:", names[int(tf.argmax(runs))])
print("Fastest scorer:", names[int(tf.argmax(strike_rate))])
print("Most sixes:", names[int(tf.argmax(sixes))])
```

A chart (chapter 10). `bottom=` stacks the orange bar on top of the grey one, and `plt.legend()` shows the colour key:

```python
plt.bar(names, running_runs.numpy(), color="#94a3b8", label="Running")
plt.bar(names, boundary_runs.numpy(), bottom=running_runs.numpy(), color="#ea580c", label="Fours and sixes")
plt.title("Where the runs came from")
plt.ylabel("Runs")
plt.legend()
plt.show()
```
