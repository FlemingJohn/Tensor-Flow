| Word | What it is | Cricket idea | Example |
|---|---|---|---|
| **Library** | A big collection of code someone else wrote | The whole kit bag | `tensorflow`, `matplotlib` |
| **Module** | One part inside a library | One pocket in the bag | `pyplot` |
| **Function** | A tool that does one job when you call it | The bat: you swing it, something happens | `plt.bar( )`, `tf.constant( )` |
| **Class** | A blueprint that makes objects | The jersey design: every player's shirt is made from it | `tf.Variable` |

So: **`matplotlib` is a library, `pyplot` is a module (not a class, not a function), and `bar` is a function inside it.**

The dot `.` means "go inside". `plt.bar` = go inside `plt` and get `bar`.

You don't have to guess. Python can tell you what something is:

```python
import inspect

print(inspect.ismodule(plt))
print(inspect.isfunction(plt.bar))
print(inspect.isclass(tf.Variable))
```

Three `True`s: `plt` is a module, `plt.bar` is a function, `tf.Variable` is a class.

`inspect` is a small toolbox that comes with Python. It's only used here to check, you won't need it again.

|---|---|---|
| 0 | `()` | One score |
| 1 | `(5,)` | One match, 5 players |
| 2 | `(5, 3)` | 5 players across 3 matches (a table) |
| 3 | `(2, 5, 3)` | 2 series, each a table |

```python
one_score = tf.constant(82)
one_match = tf.constant([82, 65, 48, 39, 27])
series = tf.constant([
    [82, 45, 101],
    [65, 12, 77],
    [48, 90, 33],
    [39, 21, 58],
    [27, 66, 14]
])
```

```python
print(one_score.shape, one_score.ndim)
print(one_match.shape, one_match.ndim)
print(series.shape, series.ndim)
```

`ndim` = the rank (*number of dimensions*). Each pair of `[ ]` brackets adds one more direction.

Now a second series, and stack both into a rank 3 tensor:

```python
series_2 = tf.constant([
    [12, 70, 55],
    [88, 34, 9],
    [61, 47, 120],
    [5, 92, 30],
    [44, 18, 71]
])
tournament = tf.stack([series, series_2])
print(tournament.shape)
```

`tf.stack` puts tensors of the same shape on top of each other, like stacking two scorecards. `(2, 5, 3)` = 2 series, 5 players, 3 matches.

`tf.Variable` is a **class** (the blueprint from chapter 2). `score = tf.Variable(0)` makes one object from that blueprint.

Later, when TensorFlow *learns*, the things it learns are stored in Variables, because they keep getting updated.

```python
score = tf.Variable(0)
score.assign_add(4)
score.assign_add(6)
score.assign_add(1)
print(score.numpy())
```

```python
score.assign(0)
print(score.numpy())
```

Runs over by over. A `for` loop repeats the indented lines once for each number in the list:

```python
overs = [8, 12, 4, 15, 9]
total = tf.Variable(0)
for runs_in_over in overs:
    total.assign_add(runs_in_over)
    print(total.numpy())
```

```python
names = ["Rohit", "Gill", "Kohli", "Iyer", "Rahul", "Hardik"]
card = tf.constant([
    [74, 58, 8, 3],
    [52, 49, 6, 1],
    [103, 87, 10, 2],
    [44, 40, 4, 1],
    [31, 27, 2, 1],
    [38, 20, 3, 3]
], dtype=tf.float32)
print(card.shape)
```

`dtype=tf.float32` makes everything decimals from the start, so you never need `tf.cast` later.

Each column is one stat. Slice them out (chapter 6):

```python
runs = card[:, 0]
balls = card[:, 1]
fours = card[:, 2]
sixes = card[:, 3]
```

Whole-tensor math (chapter 7). A four is worth 4 runs and a six is worth 6, the rest was running between the wickets:

```python
strike_rate = runs / balls * 100
boundary_runs = fours * 4 + sixes * 6
running_runs = runs - boundary_runs
print(strike_rate)
```

Summing up (chapter 8):

```python
print("Team total:", tf.reduce_sum(runs).numpy())
print("Top scorer:", names[int(tf.argmax(runs))])
print("Fastest scorer:", names[int(tf.argmax(strike_rate))])
print("Most sixes:", names[int(tf.argmax(sixes))])
```

A chart (chapter 10). `bottom=` stacks the orange bar on top of the grey one, and `plt.legend()` shows the colour key:

```python
plt.bar(names, running_runs.numpy(), color="#94a3b8", label="Running")
plt.bar(names, boundary_runs.numpy(), bottom=running_runs.numpy(), color="#ea580c", label="Fours and sixes")
plt.title("Where the runs came from")
plt.ylabel("Runs")
plt.legend()
plt.show()
```

## Words you learned

| Word | Meaning |
|---|---|
| `import ... as ...` | Bring a toolbox in and give it a nickname |
| `from ... import ...` | Take just one part out of a toolbox |
| Library / module | Whole toolbox / one drawer in it |
| Function | A tool that does one job: `name( )` |
| Class | A blueprint that makes objects |
| Tensor | A box of numbers |
| `tf.constant` | Make a locked tensor |
| `tf.Variable` | Make a tensor that can change |
| shape | How many numbers and how they're arranged |
| dtype | Whole numbers (int) or decimals (float) |
| rank / `ndim` | How many directions the numbers go |
| `[i]`, `[a:b]`, `[:, j]` | Pick one, pick a range, pick a column |
| `reduce_sum`, `reduce_mean`, `reduce_max` | Total, average, biggest |
| `argmax` | Where the biggest one is |
| `axis` | Which direction to squeeze |
| `.numpy()` | Get plain numbers out of a tensor |
