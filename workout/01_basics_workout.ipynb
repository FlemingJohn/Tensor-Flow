```python
runs = tf.constant([82, 65, 48, 39, 27])
print(runs)
```

```python
average = tf.constant([53.5, 41.2, 38.9])
print(average.dtype)
```

```python
bonus = tf.constant([53.5, 41.2, 38.9])
print(runs + bonus)
```

```python
runs_float = tf.cast(runs, tf.float32)
```

```python
one_score = tf.constant(82)
one_match = tf.constant([82, 65, 48, 39, 27])
series = tf.constant([[82, 45, 101],
    [65, 12, 77],
    [48, 90, 33],
    [39, 21, 58],
    [27, 66, 14]
    ])
```

```python
print(one_score.shape, one_score.ndim)
print(one_match.shape, one_match.ndim)
print(series.shape, series.ndim)
```

```python
series_2 = tf.constant([[12, 70, 55],
    [88, 34, 9],
    [61, 47, 120],
    [5, 92, 30],
    [44, 18, 71]
    ])
tournament = tf.stack([series, series_2])
print(tournament.shape,tournament.ndim)
```

```python
balls = tf.constant([60, 50, 40, 25, 18])
strike_rate = runs / balls * 100
print(strike_rate.numpy())
```

```python
wrestlers = ["John Cena", "Roman Reigns", "Undertaker", "Brock Lesnar", "Rey Mysterio"]
weight_kg = tf.constant([114.0, 120.0, 136.0, 130.0, 79.0])
weight_lb = weight_kg *2.2
print(weight_lb)
```

```python
print(tf.reduce_sum(runs).numpy())
print(tf.reduce_mean(runs).numpy())
print(tf.reduce_max(runs).numpy())
print(tf.reduce_min(runs).numpy())
print(tf.reduce_mean(runs_float).numpy())
```

```python
top = int(tf.argmax(runs))
print(top, players[top])
```

```python
heaviest = int(tf.argmax(weight_kg))
print(heaviest, wrestlers[heaviest])
```

```python
print(tf.reduce_sum(series, axis=1).numpy())
```

```python
print(tf.reduce_sum(series, axis=0).numpy())
```

```python
score = tf.Variable(4)
print(score.numpy())
```

```python
score.assign_add(6)
print(score.numpy())

```

```python
score.assign_sub(5)
print(score.numpy())
```

```python
score.assign(0)
print(score.numpy())
```

```python
overs = [8, 12, 4, 15, 9]
total = tf.Variable(0)
for runs_in_over in overs:
    total.assign_add(runs_in_over)
    print(total.numpy())
```

```python
plt.bar(players, strike_rate.numpy(), color='blue')
plt.title("Strike Rate")
plt.xlabel("Players")
plt.ylabel("Runs per 100 balls")
plt.show()
```

```python
fastest = int(tf.argmax(strike_rate))
colors = ['orange'] * len(players)
colors[fastest] = 'red'

plt.bar(players, strike_rate.numpy(), color=colors)
plt.title("Fastest scorer: " + players[fastest])
plt.ylabel("Runs per 100 balls")
plt.show()
```

```python
plt.barh(wrestlers, weight_kg.numpy(), color='green')
plt.title("WWE Wrestlers Weight")
plt.xlabel("Kilograms")
plt.show()
```
